## 2. An image is a tensor

A colour image is commonly represented as `(height, width, channels)`. A batch adds a leading dimension: `(batch, height, width, channels)`.

```python
image = np.zeros((5, 6, 3), dtype=np.uint8)
image[:, :2] = [255, 0, 0]
image[:, 2:4] = [0, 255, 0]
image[:, 4:] = [0, 0, 255]

print("Shape:", image.shape)
print("Data type:", image.dtype)
print("Pixel [0, 0]:", image[0, 0])
plt.imshow(image)
plt.title("Synthetic RGB image")
plt.axis("off");
```

## 3. CIFAR-10 dataset structure

```python
import tensorflow as tf

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()
class_names = ["airplane", "automobile", "bird", "cat", "deer",
               "dog", "frog", "horse", "ship", "truck"]

print("Training images:", x_train.shape, x_train.dtype)
print("Training labels:", y_train.shape, y_train.dtype)
print("Test images:", x_test.shape)
print("Test labels:", y_test.shape)
```

```python
fig, axes = plt.subplots(2, 5, figsize=(11, 5))
for i, ax in enumerate(axes.flat):
    ax.imshow(x_train[i])
    ax.set_title(class_names[int(y_train[i, 0])])
    ax.axis("off")
plt.tight_layout()
```

## 4. Normalization

CIFAR-10 pixels are stored as integers from 0 to 255. Scaling to 0–1 changes the numerical range, not the image content.

```python
x_train_float = x_train.astype("float32") / 255.0
x_test_float = x_test.astype("float32") / 255.0
print("Before:", x_train.min(), x_train.max(), x_train.dtype)
print("After :", x_train_float.min(), x_train_float.max(), x_train_float.dtype)
```

## 5. Dataset roles

- Training data: updates model parameters.
- Validation data: monitors generalization and helps make development decisions.
- Test data: held back for final evaluation.

Do not repeatedly tune against the test set.

## 6. Parameter counting

A neuron with `n` inputs has `n` weights and one bias.

For Conv2D with bias:
`(kernel_height × kernel_width × input_channels + 1) × filters`

For Dense:
`input_units × output_units + output_units`.

```python
print("Example neuron with 3 inputs:", 3 + 1, "parameters")
print("Conv2D example:", (3 * 3 * 3 + 1) * 32)  # 896
print("Dense example:", 2304 * 64 + 64)          # 147,520
```
