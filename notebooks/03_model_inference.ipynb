# 03 — Model Inference and Error Inspection
## What happens after training?

Training changes model weights. Inference uses the learned weights to predict on new images; it does not update them.

## 1. Load the saved model

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

model = tf.keras.models.load_model("cifar10_cnn.keras")
model.summary()
```

## 2. Load unseen test images

```python
(_, _), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()
class_names = ["airplane", "automobile", "bird", "cat", "deer",
               "dog", "frog", "horse", "ship", "truck"]
x_test = x_test.astype("float32") / 255.0
print("Test tensor:", x_test.shape)
```

## 3. Make a prediction

```python
sample = x_test[0]
input_batch = np.expand_dims(sample, axis=0)

scores = model.predict(input_batch, verbose=0)[0]
predicted_id = int(np.argmax(scores))
actual_id = int(y_test[0, 0])

print("Predicted:", class_names[predicted_id])
print("Actual:", class_names[actual_id])
print("Output shape:", scores.shape)
print("Sum of softmax scores:", scores.sum())

plt.imshow(sample)
plt.title(f"Predicted: {class_names[predicted_id]} | Actual: {class_names[actual_id]}")
plt.axis("off");
```

## 4. Inspect several predictions

A metric summarizes performance. Individual examples help us understand failure modes.

```python
scores = model.predict(x_test[:20], verbose=0)
predicted = np.argmax(scores, axis=1)
actual = y_test[:20].reshape(-1)

fig, axes = plt.subplots(4, 5, figsize=(12, 9))
for i, ax in enumerate(axes.flat):
    ax.imshow(x_test[i])
    ax.set_title(f"P: {class_names[predicted[i]]}\nT: {class_names[actual[i]]}",
                 color="green" if predicted[i] == actual[i] else "red", fontsize=8)
    ax.axis("off")
plt.tight_layout()
```

## 5. What do the output scores mean?

The final Dense layer has 10 units because CIFAR-10 has 10 classes. Softmax turns logits into values between 0 and 1 that sum to 1. `argmax` selects the largest score.

A high score is not a guarantee of correctness. Neural-network confidence can be miscalibrated.
